For my final project, I'm interested in exploring public transit datasets pertaining to the 1 Line Sound Transit.

# **Motivation & Problem** 

I’m interested in this analysis because I use public transit as my form of transportation to university every week. As Sound Transit continues to expand throughout Washington, I think it’s important to recognize the practical use of public transit and the various ways of adapting it. This could be the location of stations or what exists around the station, or possibly the safety by looking at the crime rate associated with each light rail station. I'm also curious about the usage of the light rail and the most populated stations.

What I'm hoping to learn from this is where SDOT plans on creating new light rail stations in future years for my personal use and knowledge about where I might want to work or live. Moreover, I'd like to know this information now for my use of the light rail, as I want to anticipate where and when riders are likely to use Sound Transit.
