# **Overview / Research Questions / Hypothesis** 

This project examines which Sound Transit 1 Line stations have grown the fastest in ridership since the Northgate extension opened in October 2021. Using Sound Transit’s monthly ridership dataset, which dates back to 2019 and is created from Automated Passenger Counter (APC) sensors near the doors of trains and buses, I will compare changes in total and average ridership across stations. As a weekly light rail commuter to the University of Washington, I want to understand where and when riders are most likely to use the system, and what that growth says about how the 1 Line is being used. The findings could also be useful to riders, planners, and researchers thinking about future extensions, and may offer a window into why stations were placed where they are and where new ones might appear next. Overall, the project seeks to turn ridership trends into practical insight for both everyday travel decisions and future transit planning.

**Research Question 1**: Do stations in dense, urban areas grow differently than park-and-ride or suburban stations?
- This question aims to compare ridership growth between urban stations (such as Capitol Hill and Westlake) and park-and-ride or suburban stations (such as Angle Lake and Tukwila Intl Blvd) to see whether a station's surroundings, such as how many people work and live nearby, affect how quickly a station grows. The answer could show planners which kinds of station locations hold up best, and help me understand why stations were placed where they are.

**Research Question 2**: How much of the post-pandemic growth is real expansion, versus recovery from the COVID-era ridership drop?
- This question aims to separate growth that comes from the system actually attracting new riders from growth that is simply riders returning after the pandemic. Because October 2021 was still a reduced baseline, this ensures my “fastest growing” conclusions to be honest by showing which stations grew faster than the pandemic recovery.

My prediction for this project is that the ridership numbers for the 1 Line will have doubled since the addition of the Northgate stations due to the increased physical accessibility of Sound Transit, rising population numbers, and return to commuting post-pandemic. 

# **Background & Related Work**

Seattle has had a long history with the light rail, as it opened on July 18, 2009 [1]. Understanding why the light rail exists in the first place is important step prior to examining ridership. A thesis paper by J. Novsam emphasizes how Seattle's transit exists because of long-standing communal travel, environmental respect, innovation, and public involvement [2]. Novsam explains this through Seattle's geography in the 1800s (where we used mostly water transport), the use of streetcars and trolleys in the early 1900s, the use of cars in the mid 1900s, and finally the Link Light Rail in the 21st century. The author further points out the placement of light rail stations to be chosen by regional planners and based around employment centers. Capitol Hill is an area of dense population, and places like Bellevue or Redmond (which the 2 Line has extended to now) are the centers of major companies such as Microsoft. The paper mentions how our community also plays an important factor in creating Sound Transit stations, as people here prioritize walkability, support voter-approved taxes, and care for the environment.

Several studies of Seattle help explain how COVID-19 changed transit ridership and what shaped people’s decisions to travel. Research on the Urban Villages program found that residents of dense neighborhoods with coordinated transit investment much more likely to ride transit or the bus, and this finding held in 2021 even though overall ridership dipped from earlier years [3]. The Seattle Flu Study, which paired respiratory virus testing with cell phone location data, showed that foot traffic to transit stations dropped more than 90% below baseline by mid March 2020. And that change, including visitors from outside the county, stayed well below normal into 2021 and did not fully recover by mid-2022. It also pointed to in-person school instruction and returning regional travel as factors that helped bring activity back [4]. Grover’s survey of Seattle-area residents suggests that fear of COVID transmission may have slowed this recovery, since people who were more concerned about infection were less supportive of higher density and expanded transit access, especially single-family homeowners [5]. All things considered, these papers suggest that ridership recovery after 2021 reflects not just a return to normal but also neighborhood density and transit access, school and work reopening, and public perceptions of risk. This matters for my project because it means stations with different surroundings may have recovered at different rates, so growth since October 2021 should be interpreted as a mix of recovery and real expansion.

# **Methodology**

This project uses a mixed-methods approach that combines quantitative analysis of Sound Transit’s ridership data with qualitative research on the context behind the numbers. For the quantitative portion, I will use Python and pandas to clean and filter the monthly APC ridership data for 1 Line stations. I will then calculate changes in average weekday and Saturday/Sunday boardings since October 2021, measured in both absolute (riders added) and relative (percent change) terms. Comparing each station’s growth to the system-wide trend will help separate real expansion from the broader post-pandemic recovery. I will present these results with tables that rank stations by growth and with charts, such as line graphs of ridership over time and bar charts comparing weekday and weekend patterns, so that trends and differences between stations are easy to see. Working in notebooks keeps the analysis transparent and reproducible, since every step from raw data to final figure is documented.

The qualitative portion fills in what the numbers cannot explain on their own. Ridership data shows which stations grew fastest, but not why. To interpret the results, I will draw on existing research from existing research on Seattle’s transit history and COVID-era travel behavior. This includes the Urban Villages study, the Seattle Flu Study mobility findings, and Grover’s survey on public attitudes toward density and transit. I will use these sources to ask why growth differs between dense urban stations and park-and-ride or suburban stations, how factors like school reopening, remote work, and new extensions may have shaped ridership, and why stations were placed where they are. A mixed-methods design fits this project because the quantitative analysis identifies patterns in the ridership data, and the qualitative research makes those patterns meaningful, so the findings are useful to riders, planners, and researchers rather than just a list of numbers.
