# **Motivation & Problem** 

For my project, the question I will be exploring is: "Which Sound Transit 1 Line stations have grown the fastest since the Northgate extension opened in October of 2021?"

I’m interested in this analysis because I use public transit as my form of transportation to university every week. As Sound Transit continues to expand throughout Washington, I think it’s important to recognize the practical use of public transit and the various ways of adapting it. Furthermore, determining what stations are most populated is not only informative to general riders, but also to engineers, designers, and researchers who are making plans to extend the 1 Line even further throughout the state.

What I'm hoping to learn from this is where Sound Transit plans on creating new light rail stations in future years for my personal use and knowledge about where I might want to work or live. Moreover, I'd like to know this information now for my use of the light rail, as I want to anticipate where and when riders are likely to use Sound Transit. I believe it's also important to recognize the history of the 1 Line and understand why stations were created where they are.

# **Data** 

Here is the link to the dataset I will be using: https://www.soundtransit.org/ride-with-us/system-performance-tracker/ridership

This dataset provides monthly ridership numbers for each mode of transportation offered by Sound Transit, and dates back to 2019. This data will be useful for my project and analysis as I'm interested specifically in the change of ridership over the years. It also provides both the total and the average ridership of each month, which will all be helpful in my analysis portion of the dataset. # **Unknowns & Dependencies** 

The dataset I'm exploring for my project utilizes Microsoft Power BI, and it is not explicit about whether the data is downloadable or not. This means if I want to create graphs and tables in my Jupyter Notebook, I'll have to manually create the tables for me to use. 

Sound Transit also provides a disclaimer in their data source notes about the APC sensors: "As of August 2022, about two-thirds of Link light rail cars have functioning APCs; data for the remaining cars is interpolated using an FTA-approved methodology." Although the interpolation methodology is approved by the FTA, this suggests the numbers in the dataset are mostly accurate, but do not have high precision (i.e., ridership numbers in the dataset are not 100% reflective of actual ridership).

Lastly, on Sound Transit's terms of use/licensing page, it says that "Any other use, including the reproduction, modification, distribution, transmission, republication, framing, display or performance of Website Content, without prior permission of Sound Transit, is strictly prohibited." This part of the licensing emphasizes how I cannot reproduce or create a display of the ridership content, meaning I should be careful with my analysis of the dataset, and it's very unlikely I will include this project on a resume or in a portfolio.
